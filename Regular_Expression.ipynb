# Returns a match where one of the specified characters (a, b, or c) are present

```python
str2 = 'Rat Cat Pat Mat Sat,Hat'
data = re.findall('[RCPM]at',str2)
print(data)
               
```

# [^abc] Returns a match for any character EXCEPT a, b, and c

```python
str3 = 'Rat Cat Pat Mat Sat Hat'
data3 = re.findall('[^RCPM]at',str3)
print(data3)
```

```python
str4 = 'Rat Cat Pat Mat Sat Hat Qat'
data4 = re.findall('[^P-R]at',str4)
print(data4)
```

```python
str4 = 'Rat Cat Pat Mat Sat Hat Qat'
reg = re.compile('[R]at')
str4=reg.sub('LION',str4)
print(str4)
```

```python
str5 = '''How are you
I am fine
Thank you'''
print(str5)
reg = re.compile('\n')
str5 = reg.sub(' ',str5)
print(str5)
```
